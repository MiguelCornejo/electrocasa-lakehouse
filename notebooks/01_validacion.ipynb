```python
# Databricks notebook source
catalog = "electrocasa"

silver_tables = [
    "ventas", "catalogo", "empleados",
    "devoluciones", "resenas", "sucursales",
    "tracking_envios"
]

gold_tables = [
    "ventas_netas_sucursal_mes",
    "top_productos_unidades",
    "top_productos_tasa_devolucion",
    "empleados_activos_sucursal_mes",
    "ventas_por_vendedor_sucursal_mes",
    "resenas_negativas_categoria_mes",
    "despachos_por_courier"
]

quarantine_tables = [
    "cuarentena_ventas",
    "cuarentena_catalogo",
    "cuarentena_empleados",
    "cuarentena_devoluciones",
    "cuarentena_resenas",
    "cuarentena_sucursales",
    "cuarentena_tracking_envios"
]

def count_table(schema, table):
    return spark.table(f"{catalog}.{schema}.{table}").count()

print("=== SILVER ===")
for table in silver_tables:
    print(f"{table}: {count_table('silver', table)}")

print("\n=== GOLD ===")
empty_gold = []

for table in gold_tables:
    total = count_table("gold", table)
    print(f"{table}: {total}")
    if total == 0:
        empty_gold.append(table)

print("\n=== CUARENTENA ===")
for table in quarantine_tables:
    print(f"{table}: {count_table('silver', table)}")

if empty_gold:
    raise Exception(
        f"Validación fallida. Tablas Gold vacías: {empty_gold}"
    )

print("\nVALIDACIÓN COMPLETADA CORRECTAMENTE")
```
